```python
# Architecture of LSTM with Keras
import tensorflow as tf
from tensorflow.keras import layers, Sequential

vocab_size = 1000      # Size of vocabulary (unique tokens)
embedding_dim = 16    # Dimension of dense embedding vectors
sequence_length = 8   # Sequence length per input sample

# Model Architecture of LSTM Classifier
lstm_model = Sequential([
    # Explicit input shape definition for modern Keras/TF
    layers.Input(shape=(sequence_length,)),
    
    # Layer 1: Word Embeddings (maps integer token IDs to 16-dim dense vectors)
    layers.Embedding(input_dim=vocab_size, output_dim=embedding_dim, name="embedding_layer"),
    
    # Layer 2: Sequence Processing with LSTM (32 memory units)
    layers.LSTM(units=32, name="lstm_layer"),
    
    # Layer 3: Binary classification output (sigmoid activation)
    layers.Dense(units=1, activation="sigmoid", name="output_layer")
], name="Sequential_LSTM_Classifier")

# Verify model architecture and parameters
lstm_model.summary()
```
