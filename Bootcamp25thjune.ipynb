## Machine Learning Fundamentals
Machine Learning models improve by measuring error (loss) and iteratively correcting parameters step by step.


### Core Problem Types:
- **Classification:** Predicting discrete categories/classes (e.g., Spam vs Not Spam, Cat vs Dog).
- **Regression:** Predicting continuous numerical values (e.g., House Prices, Stock Value, Temperature).


### Examples & Classification vs Regression:
- **Face Unlock:** Classification (Authorized vs Unauthorized)
- **Predicting Land / House Prices:** Regression (Continuous monetary amount)
- **Google Photos Face Grouping:** Classification / Clustering (Identifying specific faces)
- **Weather Temperature Forecast:** Regression (Continuous degree value)
