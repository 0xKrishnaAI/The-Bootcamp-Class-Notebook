```
# Student Performance Evaluation & Fraud Transaction Detection

# 1. Student Evaluation System
# Pre-populated sample data (or user input supported)
sample_students = {
    "Aarav": {"study_hours": 6, "attendance": 85, "marks": 82},
    "Diya": {"study_hours": 8, "attendance": 92, "marks": 91},
    "Rohan": {"study_hours": 2, "attendance": 68, "marks": 45},
    "Dev": {"study_hours": 1, "attendance": 60, "marks": 35},
    "Yash": {"study_hours": 3, "attendance": 65, "marks": 18}
}

def evaluate_student(name, data):
    print(f"\n--- Evaluation for {name} ---")
    marks = data["marks"]
    attendance = data["attendance"]
    study = data["study_hours"]
    
    if marks > 75:
        print("Performance: Strong")
    elif marks >= 40:
        print("Performance: Needs Improvement")
    else:
        print("Performance: Fail Repeat Class")
        
    if attendance < 75:
        print("Warning: Low Attendance")
        
    if study < 4:
        print("Warning: Needs More Study Time")

# Evaluate all students in dataset
for s_name, s_data in sample_students.items():
    evaluate_student(s_name, s_data)

# 2. Transaction Fraud Detection
transactions = [
    {"transaction_id": "TX101", "amount": 6200, "method": "Credit Card"},
    {"transaction_id": "TX102", "amount": 5500, "method": "Gift Card"},
    {"transaction_id": "TX103", "amount": 1200, "method": "Gift Card"},
    {"transaction_id": "TX104", "amount": 8000, "method": "Gift Card"},
    {"transaction_id": "TX105", "amount": 450, "method": "PayPal"}
]

# Flag high-risk transactions: Gift Card payments exceeding 5000
flagged_transactions = [
    t["transaction_id"] for t in transactions 
    if t["amount"] > 5000 and t["method"] == "Gift Card"
]

print("\nFlagged High-Risk Transactions:", flagged_transactions)
```
